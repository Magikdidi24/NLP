```python
import nltk
nltk.download('stopwords')
import re
import spacy

from nltk.util import ngrams
from nltk.stem.snowball import FrenchStemmer
```

# Définition du corpus

```python
corpus = [
    "J'aime le camembert et le morbier !",
    "Les chevaux du champs d'à côté sont blancs.",
    "Vous continuez de progresser tous les jours."
]
```

# Tokenisation

Tokeniser les phrases du corpus sur les espaces.

```python
res=[]
for i in corpus:
    res.append(i.split(" "))
print(res)
```

Utiliser le tokenizer de Spacy pour tokeniser les phrases du corpus.

```python
spacy_model = spacy.load("fr_core_news_sm")
```

```python
spacy_corpus = [spacy_model(phrase) for phrase in corpus]
print(spacy_corpus)
```

```python
for doc in spacy_corpus:
    for token in doc:
        print(token.text)
```

```python
corpus_tokenized = []

for doc in spacy_corpus:
    corpus_tokenized.append([token.text for token in doc])

print(corpus_tokenized)
```

# N-grammes

Afficher les unigrammes du corpus.

```python
for phrase in corpus_tokenized:
    print(list(ngrams(phrase, 1)))
```

Afficher les bigrammes du corpus.

```python
for phrase in corpus_tokenized:
    print(list(ngrams(phrase, 2)))
```

Afficher les trigrammes du corpus.

```python
for phrase in corpus_tokenized:
    print(list(ngrams(phrase, 3)))
```

# Stop words

Afficher la liste des stop words proposée par spacy.

```python
for doc in spacy_model.Defaults.stop_words:
    print(doc)
```

```python
stop_word= []
for doc in spacy_corpus:
    for token in doc:
        if token.is_stop:
            print(token.text)
            stop_word.append(token.text)
```

Ajouter des mots à la liste des stop words si besoin.

```python
for i in stop_word:
    spacy_model.Defaults.stop_words.add(i)
```

Afficher les phrases du corpus sans les stop words.

```python
a = [" ".join(token.text for token in doc if token.text not in spacy_model.Defaults.stop_words) for doc in spacy_corpus]
print(a)
```

# Stemming

```python
stemmer = FrenchStemmer()
res=[]
for i in corpus:
    res.append(i.split(" "))
for tokens in res:
    for token in tokens:
        print(token + ' --> ' + stemmer.stem(token))
```

# Lemmatisation

Afficher les phrases du corpus lemmatisées.

```python
for doc in spacy_corpus:
    for word in doc:
        print(word.text + '  ===>', word.lemma_)
```

# POS tagging

Afficher les tags pour les phrases du corpus.

```python
for doc in spacy_corpus:
    for word in doc:
        print((word.text, word.tag_))
```

# Regex

Extraire une adresse mail.

```python
mail = "Mon adresse mai est prenom.nom@adresse-mail.fr."
```

```python
import re
resultat = re.findall(r"[\w.-]+@[\w.-]+\.\w+", mail)
print(resultat)
```

Extraire un numéro de téléphone français.

```python
tel = "Mon numéro est le 0606060606 ou bien le +33606060606."
```

```python
resultats = re.findall(r"(?:\+33|0)\d{9}", tel)
print(resultats)
```

Extraire une phrase contenant le mot "python".

```python
python = "C'est la première phrase. Dans cette phrase, il y a un python! Mais pas ici. Là le python est au milieu."
```

```python
resultats = re.findall(r"[^.!?]*python[^.!?]*[.!?]", python)
print(resultats)
```
